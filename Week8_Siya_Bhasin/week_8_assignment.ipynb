# 🧠 Single Agent Pipeline Project

## Problem Statement
Build a **Single-Agent Smart Assistant** that:
- Understands user queries
- Routes tasks based on intent
- Uses tools when required
- Returns structured JSON output

### The agent should handle:
- Math queries → Calculator Tool
- Keyword extraction → Keyword Tool
- General queries → Direct response

---
### 🛠️ What You Need to Implement
- Agent logic
- Conditional routing
- Tool integration
- Basic error handling

### 🚀 Bonus
- Improve routing
- Add logging
- Add more tools


````python
# 🛠️ TOOL 1: Calculator

def calculator(expression: str) -> str:
    """Evaluate a mathematical expression."""
    try:
        return str(eval(expression))
    except Exception:
        return "Error in calculation"
````

````python
# 🛠️ TOOL 2: Keyword Extractor

def extract_keywords(text: str) -> list:
    """Extract keywords from text."""
    try:
        words = text.split()
        keywords = list(set([w.lower() for w in words if len(w) > 4]))
        return keywords[:5]
    except Exception:
        return []
````

````python
# BONUS TOOL 3: Word Counter

import re

def word_counter(text: str) -> dict:
    """Count words, characters, and sentences in a given text."""
    try:
        word_count = len(text.split())
        char_count = len(text)
        sentence_count = len(re.findall(r'[.!?]+', text)) or 1
        return {"words": word_count, "characters": char_count, "sentences": sentence_count}
    except Exception:
        return {"words": 0, "characters": 0, "sentences": 0}
````

````python
# BONUS: Logging Setup

import logging
import sys

logger = logging.getLogger("AgentLogger")
logger.setLevel(logging.INFO)

logger.propagate = False

logger.handlers.clear()

handler = logging.StreamHandler(sys.stdout)
handler.setLevel(logging.INFO)
handler.setFormatter(logging.Formatter("%(asctime)s [%(levelname)s] %(message)s"))
logger.addHandler(handler)

print("Logger ready.")
````

## 🤖 Implement Agent Logic Below

👉 Use conditional routing:
- If query contains `"calculate"` → use calculator
- If query contains `"keywords"` → use keyword extractor
- If query contains `"count"` → use word counter *(bonus)*
- Else → fallback direct textual response

````python
# 🤖 AGENT FUNCTION

def agent(query: str):
    query_lower = query.lower()
    logger.info(f"Query received: {query}")

    # Route 1: Calculator
    if "calculate" in query_lower:
        expression = query_lower.split("calculate", 1)[1].strip()
        logger.info(f"Routing to calculator | expression: {expression}")
        result = calculator(expression)
        if result == "Error in calculation":
            logger.warning(f"Bad expression — returning error type: {expression}")
            return {"type": "error", "result": f"Invalid expression: '{expression}'. Please enter a valid math expression, e.g. 'calculate 10 + 5'."}
        return {"type": "calculation", "result": result}

    # Route 2: Keyword Extraction
    elif "keywords" in query_lower:
        text = query.split("keywords", 1)[1].strip()
        if text.lower().startswith("from "):
            text = text[5:]
        logger.info(f"Routing to keyword extractor")
        result = extract_keywords(text)
        return {"type": "keywords", "result": result}

    # Route 3: Word Counter (Bonus tool)
    elif "count" in query_lower:
        text = query.split("count", 1)[1].strip()
        logger.info(f"Routing to word counter (bonus tool)")
        result = word_counter(text)
        return {"type": "word_count", "result": result}

    # Route 4: Fallback direct textual response
    else:
        logger.info(f"Routing to fallback general response")
        return {"type": "general", "result": f"This is a general query about: '{query}'. Please use 'calculate', 'keywords', or 'count' for tool-based tasks."}
````

## 📦 Expected Output Format

```
{
  "type": "calculation / keywords / word_count / general / error",
  "result": ...
}
```

````python
# 🧪 Test Cases

queries = [
    "Calculate 20 + 5",
    "Extract keywords from Artificial Intelligence is transforming industries",
    "What is machine learning?"
]

for q in queries:
    print("Query:", q)
    print("Response:", agent(q))
    print("-" * 50)
````

````python
# 🎯 Interactive Mode

while True:
    user_input = input("Enter query (type 'exit' to stop): ")
    if user_input.lower() == "exit":
        break
    print("Response:", agent(user_input))
````
